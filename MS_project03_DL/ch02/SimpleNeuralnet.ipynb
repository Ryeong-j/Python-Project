## 입력층에서 1층으로 신호 전달

```python
x = np.array([1.0, 0.5])
w1 = np.array([[0.1,0.3,0.5],[0.2,0.4,0.6]])
b1 = np.array([0.1,0.2,0.3])

print('x=',x)
print('b1=',b1)
print('w1=',w1)

a1=np.dot(x, w1) + b1
print('x*w1+b1=', a1)
```

## 활성화 함수의 적용

```python
z1 = sigmoid(a1)
print('a1=', a1)
print('z1=',z1)
```

## 1층에서 2층으로 신호 전달

```python
w2 = np.array([[0.1,0.4], [0.2,0.5], [0.3,0.6]])
b2 = np.array([0.1,0.2])

print('b2=', b2)
print('w2=', w2)

a2 = np.dot(z1, w2) + b2
print('z1*w2+b2 =', a2)
```

## 활성화 함수의 적용


```python
z2 = sigmoid(a2)
print('a2=', a2)
print('z2=', z2)
```

## 2층에서 3층으로 신호 전달

```python
w3 = np.array([[0.2, 0.5], [0.3, 0.6]])
b3 = np.array([0.1, 0.2])

print('b3 = ', b3)
print('w3 = ', w3)

a3 = np.dot(z2, w3) + b3
print('z2 * w3 + b3 =', a3)
```

## 최종 출력층에 활성화 함수로 Softmax 적용

```python
y = softmax(a3)
print('a3=', a3)
print('y=', y)
```
