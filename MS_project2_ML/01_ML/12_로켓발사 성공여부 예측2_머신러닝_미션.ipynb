```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import koreanize_matplotlib
```

#### 데이터 준비

```python
df = pd.read_csv('data/RocketLaunchDataCompleted2.csv', encoding='cp949')
df.head()
```

```python
df.info()
```

### 독립변수, 종속변수 지정

```python
X = df.drop(['Launched?', 'target'], axis=1)
y = df['target']
```

```python
# -----------------
# 데이터 유형 파악
# -----------------
cat_cols = X.select_dtypes(include='object').columns
num_cols = X.select_dtypes(include='number').columns
```

### 훈련/테스트세트 분할

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, stratify=y, random_state=42)

X_train.shape, X_test.shape, y_train.shape, y_test.shape
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# train/test 클래스 분포 비율
train_dist = y_train.value_counts(normalize=True).sort_index()
test_dist = y_test.value_counts(normalize=True).sort_index()

# 그래프 그리기
plt.figure(figsize=(10,5))
bar_width = 0.35
index = train_dist.index

plt.bar(index - bar_width/2, train_dist.values, bar_width, label='Train')
plt.bar(index + bar_width/2, test_dist.values, bar_width, label='Test')

plt.xlabel('Class')
plt.ylabel('Proportion')
plt.title('Class Distribution in Train and Test Sets')
plt.xticks(index)
plt.legend()
plt.show()
```

```python
X_train.shape, X_test.shape, y_train.shape, y_test.shape
```

### 데이터 전처리

```python
# ------------
# 스케일링
# ------------
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()

train_scaled = scaler.fit_transform(X_train[num_cols])
test_scaled = scaler.transform(X_test[num_cols])

X_train_scaled = pd.DataFrame(train_scaled, columns=num_cols)
X_test_scaled = pd.DataFrame(test_scaled, columns=num_cols)
```

```python
# ------------
# 인코딩
# ------------
from sklearn.preprocessing import OneHotEncoder
encoder = OneHotEncoder(handle_unknown='ignore', sparse_output=False, drop='first')

train_encoded = encoder.fit_transform(X_train[cat_cols])
test_encoded = encoder.transform(X_test[cat_cols])

col_names = encoder.get_feature_names_out(cat_cols)
X_train_encoded = pd.DataFrame(train_encoded, columns=col_names)
X_test_encoded = pd.DataFrame(test_encoded, columns=col_names)
```

```python
# ------------
# 최종
# ------------
X_train_preprocessed = pd.concat([X_train_scaled, X_train_encoded], axis=1)
X_test_preprocessed = pd.concat([X_test_scaled, X_test_encoded], axis=1)

X_train_preprocessed
```

### 모델 생성

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import GridSearchCV, StratifiedKFold
from sklearn.metrics import f1_score, make_scorer

# 모델과 파라미터 후보
rf = RandomForestClassifier(random_state=42)
param_grid = {
    'n_estimators': [100, 200],
    'max_depth': [None, 10, 20],
    'min_samples_split': [2, 5]
}

# StratifiedKFold 사용 (불균형 고려)
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

# f1_score 기반 평가 (불균형 대응)
scorer = make_scorer(f1_score)

grid_search = GridSearchCV(rf, param_grid, cv=cv, scoring=scorer, n_jobs=-1)
grid_search.fit(X_train_preprocessed, y_train)

print("최적 하이퍼파라미터:", grid_search.best_params_)
print("훈련 데이터 최고 F1 점수:", grid_search.best_score_)
```

### 성능평가

```python
# 테스트 데이터 평가
best_model = grid_search.best_estimator_
y_pred = best_model.predict(X_test_preprocessed)

from sklearn.metrics import classification_report
print(classification_report(y_test, y_pred))
```

```python
import pandas as pd
from sklearn.metrics import classification_report

# 예측값과 실제값 예제
y_true = y_test  # 실제값
y_pred = best_model.predict(X_test_preprocessed)  # 예측값, 예시로 앞서 최적 모델 사용

# classification_report 생성 후 딕셔너리 형태로 변환
report_dict = classification_report(y_true, y_pred, output_dict=True)

# 딕셔너리를 데이터프레임으로 변환
report_df = pd.DataFrame(report_dict).transpose()

# 출력 또는 저장
report_df
```
