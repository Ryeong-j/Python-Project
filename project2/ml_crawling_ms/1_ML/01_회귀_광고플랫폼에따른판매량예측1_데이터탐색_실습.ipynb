# 데이터 불러오기

```python
# ----------------------------------------
# 데이터를 불러와 데이터프레임을 만들고
# 상위 5개 행을 확인합니다.
# ----------------------------------------
df = pd.read_csv('Python-Project/MS_project2/ml_crawling_ms/1_ML/data/advertising.csv')
df.head()
```

# 데이터 탐색
* 독립변수와 종속변수를 식별합니다.
* 각 변수의 자료형, 결측치 여부, 통계적 특성을 확인합니다.
* 종속변수와 독립변수의 관계를 중심으로 탐색합니다.

```python
# ----------------------------------------
# 데이터프레임 정보 확인
#   - 자료형, 결측치 여부 확인
# ----------------------------------------
df.info()
```

```python
# ----------------------------------------
# 수치형 데이터 - 요약통계 확인
# ----------------------------------------
df.describe()
```

```python
# ----------------------------------------
# 수치형 독립변수의 분포 확인
# ----------------------------------------
cols = ['TV', 'Radio', 'Newspaper']

sns.histplot(x='TV', data=df)
```

```python
# ----------------------------------------
# 수치형 독립변수의 분포 확인 => 반복문으로 코드 효율화
# ----------------------------------------
fig, ax = plt.subplots(1,3, figsize=(10, 4), sharex=True, sharey=True) # 1행 3열의 그래프로 그림
# fig: 도화지, ax: 도화지에 그려지는 그래프
# sharex, sharey는 세 변수의 축의 범위를 동일하게 맞춰줌

for i, col in enumerate(cols):
    sns.histplot(x=col, data=df, ax=ax[i])
    # ax[i].set_title(col)  # 위에 제목 추가됨(아래에 이미 있으니까 생략)
    
plt.tight_layout() # 컴퓨터 화면에 맞게 fig 크기를 조절해줌(비율 말고)
```

```python
# ----------------------------------------
# 수치형 독립변수의 분포 확인 - 박스플롯
# ----------------------------------------
sns.boxplot(df[['TV', 'Radio', 'Newspaper']])
```

```python
# ----------------------------------------
# 수치형 종속변수의 분포 확인 
# ----------------------------------------
plt.figure(figsize=(3,2))
sns.histplot(x='Sales', data=df, color='green')
pass
```

```python
# ----------------------------------------
# 수치형 종속변수의 분포 확인  - 박스플롯
# ----------------------------------------
sns.boxplot(y="Sales", data=df)
```

## 독립변수와 종속변수간의 관계 파악

```python
# ----------------------------------------
# scatterplot, regplot
# ----------------------------------------
fig, ax = plt.subplots(1,3, figsize=(10, 4)) 

for i, col in enumerate(cols):
    sns.regplot(x=col, y='Sales', data=df, ax=ax[i])
    # scatterplot은 regplot에서 회귀선만 빠진 것
    # ax[i].set_title(col)  # 위에 제목 추가됨(아래에 이미 있으니까 생략)
    
plt.tight_layout() # 컴퓨터 화면에 맞게 fig 크기를 조절해줌(비율 말고)
```
